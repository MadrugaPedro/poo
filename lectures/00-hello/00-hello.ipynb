# Olá, Java!

Este notebook verifica o ambiente e apresenta a primeira instrução Java.

## Objetivos

- executar uma célula Java;
- produzir uma mensagem na saída padrão;
- identificar a versão da JVM usada pelo notebook.

## Como executar

Execute as células de código em ordem. O kernel de notebook aceita fragmentos de Java de forma semelhante ao JShell. Em um arquivo `.java` tradicional, instruções executáveis ficam dentro de um método, normalmente `main`.

> O ponto e vírgula encerra a instrução e deve ser mantido, mesmo quando um ambiente interativo aceita sua omissão.

## Primeira demonstração

`System.out.println` escreve uma mensagem na saída padrão e acrescenta uma quebra de linha. A saída esperada é `Olá, mundo!`.

```java
System.out.println("Olá, mundo!");
```

## Ambiente de execução

A propriedade `java.version` informa a versão da JVM. O valor exibido depende do ambiente aberto no VS Code.

```java
System.out.println("Versão da JVM: " + System.getProperty("java.version"));
```
