# Controle de fluxo em Java

## 1. Condições e expressões booleanas

Uma estrutura de seleção decide qual trecho executar a partir de uma expressão do tipo `boolean`. As condições normalmente combinam:

| Categoria | Operadores | Exemplo |
|---|---|---|
| Relacionais | `<`, `<=`, `>`, `>=` | `idade >= 18` |
| Igualdade | `==`, `!=` | `tentativas != 0` |
| Lógicos | `&&`, `\|\|`, `!` | `temIngresso && !eventoEncerrado` |

Os operadores `&&` e `||` usam **curto-circuito**: a segunda parte só é avaliada quando necessária. Para comparar o conteúdo de objetos, como `String`, normalmente usamos `equals`, e não `==`.

````java
int idade = 20;
boolean possuiDocumento = true;
boolean podeEntrar = idade >= 18 && possuiDocumento;
System.out.println("Pode entrar? " + podeEntrar);
````

## 2. Seleção simples com `if`

O `if` executa um bloco **somente quando** sua condição é verdadeira. Se ela for falsa, o programa simplesmente continua após o bloco.

```java
if (condicao) {
    // executado quando a condição é verdadeira
}
```

As chaves são opcionais para uma única instrução, mas mantê-las reduz erros durante futuras alterações do código.

````java
double temperatura = 38.2;

if (temperatura >= 37.8) {
    System.out.println("Atenção: temperatura elevada.");
}
````

### Variações: vários `if` e `if` aninhado

Vários `if` independentes podem executar mais de um bloco. Um `if` aninhado permite testar uma segunda condição apenas dentro de um caminho já escolhido. Use o aninhamento com moderação: condições compostas ou retornos antecipados costumam deixar o código mais simples.

````java
int numero = 12;

if (numero > 0) {
    System.out.println("O número é positivo.");
}
if (numero % 2 == 0) {
    System.out.println("O número é par.");
}

boolean usuarioAtivo = true;
boolean senhaCorreta = true;
if (usuarioAtivo) {
    if (senhaCorreta) {
        System.out.println("Acesso autorizado.");
    }
}
````

## 3. Seleção composta com `if-else`

O `if-else` representa dois caminhos mutuamente exclusivos: exatamente um deles será executado.

```java
if (condicao) {
    // caminho verdadeiro
} else {
    // caminho falso
}
```

````java
double nota = 6.5;

if (nota >= 6.0) {
    System.out.println("Aprovado");
} else {
    System.out.println("Reprovado");
}
````

### Variação: encadeamento `if-else if-else`

O encadeamento testa alternativas em ordem. A primeira condição verdadeira vence; as seguintes não são avaliadas. Por isso, intervalos mais específicos devem aparecer antes dos mais abrangentes.

````java
double media = 8.7;
String conceito;

if (media >= 9.0) {
    conceito = "A";
} else if (media >= 7.0) {
    conceito = "B";
} else if (media >= 5.0) {
    conceito = "C";
} else {
    conceito = "D";
}
System.out.println("Conceito: " + conceito);
````

### Demonstração — classificação por faixa etária

A implementação usa `if-else if-else` para classificar uma idade como `criança` (até 11), `adolescente` (12 a 17), `adulto` (18 a 59) ou `idoso` (60 ou mais). Idades negativas são tratadas como inválidas, e os valores de fronteira são demonstrados.

````java
static String classificarFaixaEtaria(int idade) {
    if (idade < 0) {
        return "idade inválida";
    } else if (idade <= 11) {
        return "criança";
    } else if (idade <= 17) {
        return "adolescente";
    } else if (idade <= 59) {
        return "adulto";
    } else {
        return "idoso";
    }
}

for (int idadeTeste : new int[] {-1, 11, 12, 17, 18, 59, 60}) {
    System.out.printf("%d: %s%n", idadeTeste, classificarFaixaEtaria(idadeTeste));
}
````

## 4. Operador condicional ternário `?:`

O ternário é uma **expressão** que produz um valor, diferentemente do `if`, que é uma instrução. Sua forma é:

```java
condicao ? valorSeVerdadeiro : valorSeFalso
```

Ele é adequado para uma decisão curta que escolhe entre dois valores. Evite ternários longos, com efeitos colaterais ou muitos níveis de aninhamento.

````java
int estoque = 3;
String disponibilidade = estoque > 0 ? "Disponível" : "Esgotado";
System.out.println(disponibilidade);

int a = 17;
int b = 9;
int maior = a > b ? a : b;
System.out.println("Maior valor: " + maior);
````

### Variação: ternário encadeado

O ternário pode ser encadeado, mas perde legibilidade rapidamente. O exemplo abaixo é aceitável por ser pequeno; para regras maiores, prefira `if-else if-else` ou um método com nome descritivo.

````java
int valor = 0;
String sinal = valor > 0 ? "positivo" : valor < 0 ? "negativo" : "zero";
System.out.println(sinal);
````

### Demonstração — frete grátis

O operador ternário produz `"Frete grátis"` quando o total da compra é de pelo menos R$ 150,00 e `"Frete: R$ 20,00"` nos demais casos. A demonstração inclui valores abaixo, no limite e acima dele.

````java
for (double totalCompra : new double[] {149.99, 150.0, 175.0}) {
    String mensagemFrete = totalCompra >= 150.0 ? "Frete grátis" : "Frete: R$ 20,00";
    System.out.printf("Compra de R$ %.2f: %s%n", totalCompra, mensagemFrete);
}
````

## 5. `switch` tradicional

O `switch` é útil quando uma única expressão é comparada com vários valores discretos. Na forma tradicional, cada `case` usa `:` e normalmente termina com `break`. Sem o `break`, a execução continua no caso seguinte — comportamento chamado ***fall-through***. O `default` trata valores não contemplados.

Seletores tradicionais incluem tipos inteiros compatíveis (`byte`, `short`, `char` e `int` e seus *wrappers*), `String` e tipos `enum`. `long`, `float`, `double` e `boolean` não são seletores tradicionais válidos.

````java
int diaDaSemana = 3;
String nomeDoDia;

switch (diaDaSemana) {
    case 1:
        nomeDoDia = "domingo";
        break;
    case 2:
        nomeDoDia = "segunda-feira";
        break;
    case 3:
        nomeDoDia = "terça-feira";
        break;
    default:
        nomeDoDia = "outro dia";
}
System.out.println(nomeDoDia);
````

### Variação tradicional: casos agrupados por *fall-through*

Casos vazios podem compartilhar o mesmo bloco. Aqui, o *fall-through* é deliberado. Mesmo assim, a sintaxe moderna com rótulos separados por vírgula costuma comunicar melhor a intenção.

````java
char categoriaCarteira = 'B';

switch (categoriaCarteira) {
    case 'A':
    case 'B':
        System.out.println("Veículo particular");
        break;
    case 'C':
    case 'D':
    case 'E':
        System.out.println("Veículo profissional");
        break;
    default:
        System.out.println("Categoria inválida");
}
````

## 6. `switch` moderno com `->`

Rótulos com seta não produzem *fall-through* e aceitam vários valores separados por vírgula. À direita da seta podemos usar uma expressão, uma instrução ou um bloco. Essa forma é mais segura quando cada caso é independente.

````java
String mes = "FEV";

switch (mes) {
    case "DEZ", "JAN", "FEV" -> System.out.println("Verão");
    case "MAR", "ABR", "MAI" -> System.out.println("Outono");
    case "JUN", "JUL", "AGO" -> System.out.println("Inverno");
    case "SET", "OUT", "NOV" -> System.out.println("Primavera");
    default -> System.out.println("Mês inválido");
}
````

## 7. Expressão `switch`

O `switch` também pode **produzir um valor**. Quando usamos `->`, o valor aparece diretamente após a seta. Uma expressão `switch` deve ser exaustiva: todos os valores possíveis precisam ser cobertos, geralmente com `default`.

````java
int numeroDoMes = 8;
String trimestre = switch (numeroDoMes) {
    case 1, 2, 3 -> "1º trimestre";
    case 4, 5, 6 -> "2º trimestre";
    case 7, 8, 9 -> "3º trimestre";
    case 10, 11, 12 -> "4º trimestre";
    default -> "mês inválido";
};
System.out.println(trimestre);
````

### Variação: bloco e `yield`

Quando um caso precisa de várias instruções para calcular seu resultado, usamos um bloco. A palavra-chave `yield` entrega o valor daquele bloco à expressão `switch`; ela não tem a mesma função de `break`.

````java
String plano = "PREMIUM";
double mensalidade = switch (plano) {
    case "BASICO" -> 29.90;
    case "PREMIUM" -> {
        double precoBase = 59.90;
        double descontoPromocional = 5.00;
        yield precoBase - descontoPromocional;
    }
    case "FAMILIA" -> 89.90;
    default -> throw new IllegalArgumentException("Plano desconhecido: " + plano);
};
System.out.printf("Mensalidade: R$ %.2f%n", mensalidade);
````

### Variação: `switch` com `enum`

Um `enum` restringe os valores possíveis e combina muito bem com uma expressão `switch`. Quando todos os valores do `enum` são tratados, o compilador reconhece a exaustividade sem exigir `default`.

````java
enum Prioridade { BAIXA, MEDIA, ALTA }

Prioridade prioridade = Prioridade.ALTA;
int prazoEmHoras = switch (prioridade) {
    case BAIXA -> 72;
    case MEDIA -> 24;
    case ALTA -> 4;
};
System.out.println("Prazo: " + prazoEmHoras + " horas");
````

## 8. Casamento de padrões no `switch` — Java 21

O `switch` moderno pode selecionar caminhos pelo **tipo** de um objeto. A variável declarada no padrão já fica disponível no caso correspondente. Uma guarda com `when` acrescenta uma condição ao padrão. A ordem importa: casos específicos devem vir antes dos mais gerais. `case null` permite tratar a referência nula explicitamente.

````java
Object dado = -7;
String descricao = switch (dado) {
    case null -> "valor nulo";
    case Integer n when n < 0 -> "inteiro negativo: " + n;
    case Integer n -> "inteiro não negativo: " + n;
    case String texto -> "texto com " + texto.length() + " caracteres";
    default -> "outro tipo: " + dado.getClass().getSimpleName();
};
System.out.println(descricao);
````

## 9. Como escolher a estrutura?

| Situação | Estrutura sugerida |
|---|---|
| Executar algo apenas se uma condição ocorrer | `if` |
| Escolher entre dois blocos | `if-else` |
| Testar faixas, condições diferentes ou regras complexas | `if-else if-else` |
| Escolher de forma curta entre dois valores | ternário `?:` |
| Comparar um valor com várias constantes discretas | `switch` |
| Produzir um valor a partir de várias alternativas | expressão `switch` |
| Selecionar pelo tipo de uma referência | `switch` com padrões |

Prefira clareza a concisão. Se uma expressão exige releitura, divida-a em variáveis com nomes significativos ou extraia a regra para um método.

## 10. Demonstração integrada — calculadora

A calculadora combina `if` e uma expressão `switch`. Os operadores válidos são `+`, `-`, `*` e `/`; a implementação rejeita divisão por zero e operadores desconhecidos.

````java
static double calcular(double operando1, char operador, double operando2) {
    if (operador == '/' && operando2 == 0.0) {
        throw new ArithmeticException("Divisão por zero");
    }
    return switch (operador) {
        case '+' -> operando1 + operando2;
        case '-' -> operando1 - operando2;
        case '*' -> operando1 * operando2;
        case '/' -> operando1 / operando2;
        default -> throw new IllegalArgumentException("Operador inválido: " + operador);
    };
}

for (char operadorTeste : new char[] {'+', '-', '*', '/'}) {
    System.out.printf("20 %c 4 = %.1f%n",
            operadorTeste, calcular(20.0, operadorTeste, 4.0));
}
````

## 11. Demonstração integrada — política de descontos

A implementação calcula o preço final de uma compra seguindo estas regras:

1. clientes `BRONZE`, `PRATA` e `OURO` recebem, respectivamente, 0%, 5% e 10% de desconto;
2. compras a partir de R$ 500,00 recebem mais 5 pontos percentuais;
3. o desconto total não pode ultrapassar 15%;
4. valores de compra negativos são inválidos.

A demonstração combina `enum`, expressão `switch`, `if` e operador ternário e inclui casos abaixo, no limite e acima de R$ 500,00.

````java
enum CategoriaCliente { BRONZE, PRATA, OURO }

record ResumoCompra(double desconto, double precoFinal) {}

static ResumoCompra aplicarDesconto(CategoriaCliente categoria, double valorCompra) {
    if (valorCompra < 0.0) {
        throw new IllegalArgumentException("O valor da compra não pode ser negativo");
    }
    double descontoBase = switch (categoria) {
        case BRONZE -> 0.00;
        case PRATA -> 0.05;
        case OURO -> 0.10;
    };
    double bonus = valorCompra >= 500.0 ? 0.05 : 0.00;
    double descontoTotal = Math.min(descontoBase + bonus, 0.15);
    return new ResumoCompra(descontoTotal, valorCompra * (1.0 - descontoTotal));
}

for (double valorTeste : new double[] {499.99, 500.0, 600.0}) {
    ResumoCompra resumo = aplicarDesconto(CategoriaCliente.PRATA, valorTeste);
    System.out.printf("Compra R$ %.2f: desconto %.0f%%, total R$ %.2f%n",
            valorTeste, resumo.desconto() * 100, resumo.precoFinal());
}
````

## Síntese

- `if` trata uma condição; `if-else` escolhe entre dois caminhos.
- Encadeamentos `else if` são apropriados para faixas e regras diferentes.
- O ternário escolhe entre dois **valores** e deve permanecer curto.
- O `switch` tradicional exige cuidado com `break` e *fall-through*.
- Rótulos `case ... ->` eliminam o *fall-through* acidental.
- Expressões `switch` retornam valores; blocos usam `yield`.
- No Java 21, padrões permitem selecionar pelo tipo e adicionar guardas com `when`.
