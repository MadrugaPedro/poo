# Controle de fluxo em Java

Este notebook apresenta as estruturas de **controle de fluxo** da linguagem Java: seleção com `if`, `if-else`, operador ternário e `switch`; e iteração com `while`, `do-while`, `for` e `for-each`.

## Objetivos

Ao final, você deverá ser capaz de:

- construir condições que resultem em `true` ou `false`;
- escolher entre um, dois ou vários caminhos de execução;
- distinguir instruções de expressões;
- selecionar a estrutura mais adequada para cada problema;
- reconhecer e evitar problemas como condições redundantes e *fall-through* acidental;
- construir repetições controladas por condição, contador e coleção;
- usar `break` e `continue` conscientemente.

> Regras `case ... ->` e expressões `switch` são definitivas desde Java 14. O casamento de padrões, `case null` e guardas `when` requerem Java 21. Execute as células em ordem usando um kernel Java.

## 1. Condições e expressões booleanas

Uma estrutura de seleção decide qual trecho executar a partir de uma expressão do tipo `boolean`. As condições normalmente combinam:

| Categoria | Operadores | Exemplo |
|---|---|---|
| Relacionais | `<`, `<=`, `>`, `>=` | `idade >= 18` |
| Igualdade | `==`, `!=` | `tentativas != 0` |
| Lógicos | `&&`, `\|\|`, `!` | `temIngresso && !eventoEncerrado` |

Os operadores `&&` e `||` usam **curto-circuito**: a segunda parte só é avaliada quando necessária. Para comparar o conteúdo de objetos, como `String`, normalmente usamos `equals`, e não `==`.

````java
int idade = 20;
boolean possuiDocumento = true;
boolean podeEntrar = idade >= 18 && possuiDocumento;
System.out.println("Pode entrar? " + podeEntrar);
````

## 2. Seleção simples com `if`

O `if` executa um bloco **somente quando** sua condição é verdadeira. Se ela for falsa, o programa simplesmente continua após o bloco.

```java
if (condicao) {
    // executado quando a condição é verdadeira
}
```

As chaves são opcionais para uma única instrução, mas mantê-las reduz erros durante futuras alterações do código.

**Referência oficial:** [JLS §14.9 — The `if` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.9).

````java
double temperatura = 38.2;

if (temperatura >= 37.8) {
    System.out.println("Atenção: temperatura elevada.");
}
````

### Variações: vários `if` e `if` aninhado

Vários `if` independentes podem executar mais de um bloco. Um `if` aninhado permite testar uma segunda condição apenas dentro de um caminho já escolhido. Use o aninhamento com moderação: condições compostas ou retornos antecipados costumam deixar o código mais simples.

````java
int numero = 12;

if (numero > 0) {
    System.out.println("O número é positivo.");
}
if (numero % 2 == 0) {
    System.out.println("O número é par.");
}

boolean usuarioAtivo = true;
boolean senhaCorreta = true;
if (usuarioAtivo) {
    if (senhaCorreta) {
        System.out.println("Acesso autorizado.");
    }
}
````

## 3. Seleção composta com `if-else`

O `if-else` representa dois caminhos mutuamente exclusivos: exatamente um deles será executado.

```java
if (condicao) {
    // caminho verdadeiro
} else {
    // caminho falso
}
```

````java
double nota = 6.5;

if (nota >= 6.0) {
    System.out.println("Aprovado");
} else {
    System.out.println("Reprovado");
}
````

### Variação: encadeamento `if-else if-else`

O encadeamento testa alternativas em ordem. A primeira condição verdadeira vence; as seguintes não são avaliadas. Por isso, intervalos mais específicos devem aparecer antes dos mais abrangentes.

````java
double media = 8.7;
String conceito;

if (media >= 9.0) {
    conceito = "A";
} else if (media >= 7.0) {
    conceito = "B";
} else if (media >= 5.0) {
    conceito = "C";
} else {
    conceito = "D";
}
System.out.println("Conceito: " + conceito);
````

### Demonstração — classificação por faixa etária

A implementação usa `if-else if-else` para classificar uma idade como `criança` (até 11), `adolescente` (12 a 17), `adulto` (18 a 59) ou `idoso` (60 ou mais). Idades negativas são tratadas como inválidas.

````java
int idadeFaixa = 18;
String faixaEtaria;

if (idadeFaixa < 0) {
    faixaEtaria = "idade inválida";
} else if (idadeFaixa <= 11) {
    faixaEtaria = "criança";
} else if (idadeFaixa <= 17) {
    faixaEtaria = "adolescente";
} else if (idadeFaixa <= 59) {
    faixaEtaria = "adulto";
} else {
    faixaEtaria = "idoso";
}
System.out.printf("%d: %s%n", idadeFaixa, faixaEtaria);
````

## 4. Operador condicional ternário `?:`

O ternário é uma **expressão** que produz um valor, diferentemente do `if`, que é uma instrução. Sua forma é:

```java
condicao ? valorSeVerdadeiro : valorSeFalso
```

Ele é adequado para uma decisão curta que escolhe entre dois valores. Evite ternários longos, com efeitos colaterais ou muitos níveis de aninhamento.

**Referência oficial:** [JLS §15.25 — Conditional Operator `? :`](https://docs.oracle.com/javase/specs/jls/se25/html/jls-15.html#jls-15.25).

````java
int estoque = 3;
String disponibilidade = estoque > 0 ? "Disponível" : "Esgotado";
System.out.println(disponibilidade);

int a = 17;
int b = 9;
int maior = a > b ? a : b;
System.out.println("Maior valor: " + maior);
````

### Variação: ternário encadeado

O ternário pode ser encadeado, mas perde legibilidade rapidamente. O exemplo abaixo é aceitável por ser pequeno; para regras maiores, prefira `if-else if-else` ou um método com nome descritivo.

````java
int valor = 0;
String sinal = valor > 0 ? "positivo" : valor < 0 ? "negativo" : "zero";
System.out.println(sinal);
````

### Demonstração — frete grátis

O operador ternário produz `"Frete grátis"` quando o total da compra é de pelo menos R$ 150,00 e `"Frete: R$ 20,00"` nos demais casos.

````java
double totalCompraFrete = 175.0;
String mensagemFrete = totalCompraFrete >= 150.0 ? "Frete grátis" : "Frete: R$ 20,00";
System.out.printf("Compra de R$ %.2f: %s%n", totalCompraFrete, mensagemFrete);
````

## 5. `switch` tradicional

O `switch` é útil quando uma única expressão é comparada com vários valores discretos. Na forma tradicional, cada `case` usa `:` e normalmente termina com `break`. Sem o `break`, a execução continua no caso seguinte — comportamento chamado ***fall-through***. O `default` trata valores não contemplados.

Seletores tradicionais incluem tipos inteiros compatíveis (`byte`, `short`, `char` e `int` e seus *wrappers*), `String` e tipos `enum`. `long`, `float`, `double` e `boolean` não são seletores tradicionais válidos.

**Referência oficial:** [JLS §14.11 — The `switch` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.11).

````java
int diaDaSemana = 3;
String nomeDoDia;

switch (diaDaSemana) {
    case 1:
        nomeDoDia = "domingo";
        break;
    case 2:
        nomeDoDia = "segunda-feira";
        break;
    case 3:
        nomeDoDia = "terça-feira";
        break;
    default:
        nomeDoDia = "outro dia";
}
System.out.println(nomeDoDia);
````

### Variação tradicional: casos agrupados por *fall-through*

Casos vazios podem compartilhar o mesmo bloco. Aqui, o *fall-through* é deliberado. Mesmo assim, a sintaxe moderna com rótulos separados por vírgula costuma comunicar melhor a intenção.

````java
char categoriaCarteira = 'B';

switch (categoriaCarteira) {
    case 'A':
    case 'B':
        System.out.println("Veículo particular");
        break;
    case 'C':
    case 'D':
    case 'E':
        System.out.println("Veículo profissional");
        break;
    default:
        System.out.println("Categoria inválida");
}
````

## 6. `switch` moderno com `->`

Rótulos com seta não produzem *fall-through* e aceitam vários valores separados por vírgula. À direita da seta podemos usar uma expressão, uma instrução ou um bloco. Essa forma é mais segura quando cada caso é independente.

````java
String mes = "FEV";

switch (mes) {
    case "DEZ", "JAN", "FEV" -> System.out.println("Verão");
    case "MAR", "ABR", "MAI" -> System.out.println("Outono");
    case "JUN", "JUL", "AGO" -> System.out.println("Inverno");
    case "SET", "OUT", "NOV" -> System.out.println("Primavera");
    default -> System.out.println("Mês inválido");
}
````

## 7. Expressão `switch`

O `switch` também pode **produzir um valor**. Quando usamos `->`, o valor aparece diretamente após a seta. Uma expressão `switch` deve ser exaustiva: todos os valores possíveis precisam ser cobertos, geralmente com `default`.

**Referência oficial:** [JLS §15.28 — `switch` Expressions](https://docs.oracle.com/javase/specs/jls/se25/html/jls-15.html#jls-15.28).

````java
int numeroDoMes = 8;
String trimestre = switch (numeroDoMes) {
    case 1, 2, 3 -> "1º trimestre";
    case 4, 5, 6 -> "2º trimestre";
    case 7, 8, 9 -> "3º trimestre";
    case 10, 11, 12 -> "4º trimestre";
    default -> "mês inválido";
};
System.out.println(trimestre);
````

### Variação: bloco e `yield`

Quando um caso precisa de várias instruções para calcular seu resultado, usamos um bloco. A palavra-chave `yield` entrega o valor daquele bloco à expressão `switch`; ela não tem a mesma função de `break`.

````java
String plano = "PREMIUM";
double mensalidade = switch (plano) {
    case "BASICO" -> 29.90;
    case "PREMIUM" -> {
        double precoBase = 59.90;
        double descontoPromocional = 5.00;
        yield precoBase - descontoPromocional;
    }
    case "FAMILIA" -> 89.90;
    default -> throw new IllegalArgumentException("Plano desconhecido: " + plano);
};
System.out.printf("Mensalidade: R$ %.2f%n", mensalidade);
````

## 8. Conteúdo complementar: padrões no `switch` — Java 21

Esta variação antecipa tipos por referência, que serão aprofundados nas aulas de classes. O `switch` pode selecionar caminhos pelo **tipo** de um objeto. A variável declarada no padrão fica disponível no caso correspondente; uma guarda com `when` acrescenta uma condição. Casos específicos devem vir antes dos gerais, e `case null` trata a referência nula explicitamente.

**Referência oficial:** [JLS §14.11.1 — Switch Blocks](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.11.1).

````java
Object dado = -7;
String descricao = switch (dado) {
    case null -> "valor nulo";
    case Integer n when n < 0 -> "inteiro negativo: " + n;
    case Integer n -> "inteiro não negativo: " + n;
    case String texto -> "texto com " + texto.length() + " caracteres";
    default -> "outro tipo: " + dado.getClass().getSimpleName();
};
System.out.println(descricao);
````

## 9. Como escolher a estrutura?

| Situação | Estrutura sugerida |
|---|---|
| Executar algo apenas se uma condição ocorrer | `if` |
| Escolher entre dois blocos | `if-else` |
| Testar faixas, condições diferentes ou regras complexas | `if-else if-else` |
| Escolher de forma curta entre dois valores | ternário `?:` |
| Comparar um valor com várias constantes discretas | `switch` |
| Produzir um valor a partir de várias alternativas | expressão `switch` |
| Selecionar pelo tipo de uma referência | `switch` com padrões |
| Repetir enquanto uma condição for verdadeira | `while` |
| Executar antes de testar a condição | `do-while` |
| Repetir com contador ou atualização regular | `for` |
| Percorrer todos os elementos | `for-each` |

Prefira clareza a concisão. Se uma expressão exige releitura, divida-a em variáveis com nomes significativos.

## 10. Estruturas de iteração

Estruturas de iteração repetem uma instrução ou bloco. Toda repetição precisa de uma forma de progresso para que sua condição possa se tornar falsa; caso contrário, surge um laço infinito.

| Estrutura | Uso mais comum | Teste da condição |
|---|---|---|
| `while` | quantidade de repetições desconhecida | antes do corpo |
| `do-while` | corpo precisa executar ao menos uma vez | depois do corpo |
| `for` | inicialização, condição e atualização relacionadas | antes do corpo |
| `for-each` | percorrer todos os elementos de um array ou coleção | controlado pelo percurso |

**Referência oficial:** [JLS §14.12–§14.14 — Iteration Statements](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.12).

## 11. Repetição com `while`

O `while` avalia uma expressão `boolean` antes de cada repetição. Se ela for inicialmente falsa, o corpo não será executado.

```java
while (condicao) {
    // corpo
}
```

A demonstração separa claramente inicialização, condição e atualização.

**Referência oficial:** [JLS §14.12 — The `while` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.12).

````java
int contadorWhile = 1;             // inicialização
while (contadorWhile <= 5) {       // condição
    System.out.print(contadorWhile + " " );
    contadorWhile++;               // atualização
}
System.out.println();
````

## 12. Repetição com `do-while`

O `do-while` avalia a condição depois do corpo, portanto executa o bloco pelo menos uma vez. O ponto e vírgula após `while (condicao);` faz parte da sintaxe.

```java
do {
    // corpo
} while (condicao);
```

**Referência oficial:** [JLS §14.13 — The `do` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.13).

````java
int tentativaDoWhile = 1;
do {
    System.out.println("Tentativa " + tentativaDoWhile);
    tentativaDoWhile++;
} while (tentativaDoWhile <= 3);
````

## 13. Repetição com `for`

O `for` básico reúne inicialização, condição e atualização no cabeçalho. É especialmente adequado para contagens.

```java
for (inicializacao; condicao; atualizacao) {
    // corpo
}
```

A variável declarada no cabeçalho existe apenas no laço. Todas as três partes são opcionais; `for (;;)`, por exemplo, representa um laço infinito deliberado. Mais de uma inicialização ou atualização pode ser separada por vírgula.

**Referência oficial:** [JLS §14.14.1 — The Basic `for` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.14.1).

````java
int somaFor = 0;
for (int numeroFor = 1; numeroFor <= 5; numeroFor++) {
    somaFor += numeroFor;
}
System.out.println("Soma de 1 a 5: " + somaFor);

for (int esquerdaFor = 1, direitaFor = 5;
        esquerdaFor < direitaFor;
        esquerdaFor++, direitaFor--) {
    System.out.printf("%d < %d%n", esquerdaFor, direitaFor);
}
````

## 14. Percurso com `for-each`

O `for` aprimorado, conhecido como `for-each`, percorre cada elemento de um array ou de um objeto `Iterable`. Um array reúne valores do mesmo tipo; `int[]` significa “array de inteiros”.

```java
for (tipo elemento : origem) {
    // usa elemento
}
```

Ele é preferível quando o índice não é necessário. A variável `elemento` recebe, a cada repetição, o próximo valor da origem.

**Referência oficial:** [JLS §14.14.2 — The Enhanced `for` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.14.2).

````java
int[] notasForEach = {8, 7, 9, 10};
int totalForEach = 0;
for (int notaForEach : notasForEach) {
    totalForEach += notaForEach;
}
double mediaForEach = (double) totalForEach / notasForEach.length;
System.out.println("Média: " + mediaForEach);
````

## 15. Laços aninhados

Uma estrutura de iteração pode aparecer dentro de outra. Para cada repetição do laço externo, o interno executa seu ciclo completo. Isso é útil para linhas e colunas, mas aumenta rapidamente o número de operações.

````java
for (int linha = 1; linha <= 3; linha++) {
    for (int coluna = 1; coluna <= 4; coluna++) {
        System.out.print("* " );
    }
    System.out.println();
}
````

## 16. Desvios com `break` e `continue`

`break` encerra imediatamente o laço mais interno. `continue` abandona apenas a repetição atual e segue para a próxima. Ambos podem ser úteis, mas uso excessivo dificulta acompanhar o fluxo.

**Referências oficiais:** [JLS §14.15 — The `break` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.15) e [JLS §14.16 — The `continue` Statement](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.16).

````java
for (int numeroDesvio = 1; numeroDesvio <= 10; numeroDesvio++) {
    if (numeroDesvio == 3) {
        continue; // não imprime 3
    }
    if (numeroDesvio == 7) {
        break;    // encerra antes de imprimir 7
    }
    System.out.print(numeroDesvio + " " );
}
System.out.println();
````

## 17. Demonstração integrada — calculadora

A calculadora combina `if` e uma expressão `switch`. Os operadores válidos são `+`, `-`, `*` e `/`; a demonstração verifica a divisão por zero antes de calcular.

````java
double operando1 = 20.0;
double operando2 = 4.0;
char operadorCalculadora = '/';
String resultadoCalculadora;

if (operadorCalculadora == '/' && operando2 == 0.0) {
    resultadoCalculadora = "divisão por zero";
} else {
    resultadoCalculadora = switch (operadorCalculadora) {
        case '+' -> String.valueOf(operando1 + operando2);
        case '-' -> String.valueOf(operando1 - operando2);
        case '*' -> String.valueOf(operando1 * operando2);
        case '/' -> String.valueOf(operando1 / operando2);
        default -> "operador inválido";
    };
}
System.out.println("Resultado: " + resultadoCalculadora);
````

## 18. Demonstração integrada — política de descontos

A implementação calcula o preço final de uma compra seguindo estas regras:

1. categorias `BRONZE`, `PRATA` e `OURO` recebem, respectivamente, 0%, 5% e 10% de desconto;
2. compras a partir de R$ 500,00 recebem mais 5 pontos percentuais;
3. o desconto total não pode ultrapassar 15%;
4. valores de compra negativos são inválidos.

A demonstração combina uma expressão `switch`, `if` e operador ternário usando apenas variáveis e tipos já apresentados.

````java
String categoriaCliente = "PRATA";
double valorCompraDesconto = 600.0;

if (valorCompraDesconto < 0.0) {
    System.out.println("Valor de compra inválido");
} else {
    double descontoBase = switch (categoriaCliente) {
        case "BRONZE" -> 0.00;
        case "PRATA" -> 0.05;
        case "OURO" -> 0.10;
        default -> 0.00;
    };
    double bonusDesconto = valorCompraDesconto >= 500.0 ? 0.05 : 0.00;
    double descontoTotal = Math.min(descontoBase + bonusDesconto, 0.15);
    double precoFinal = valorCompraDesconto * (1.0 - descontoTotal);
    System.out.printf("Desconto: %.0f%%; total: R$ %.2f%n",
            descontoTotal * 100, precoFinal);
}
````

## Síntese

- `if` trata uma condição; `if-else` escolhe entre dois caminhos.
- Encadeamentos `else if` são apropriados para faixas e regras diferentes.
- O ternário escolhe entre dois **valores** e deve permanecer curto.
- O `switch` tradicional exige cuidado com `break` e *fall-through*.
- Rótulos `case ... ->` eliminam o *fall-through* acidental.
- Expressões `switch` retornam valores; blocos usam `yield`.
- No Java 21, padrões permitem selecionar pelo tipo e adicionar guardas com `when`.
- `while` testa a condição antes do corpo; `do-while`, depois.
- `for` reúne inicialização, condição e atualização no cabeçalho.
- `for-each` percorre arrays e objetos iteráveis sem índice explícito.
- `break` encerra o laço; `continue` avança para a próxima repetição.
