# Elementos da linguagem Java

Este notebook trata exclusivamente dos elementos fundamentais solicitados: **literais, identificadores, palavras-chave, separadores, operadores, tipos primitivos, variáveis, declarações, expressões e instruções**.

O conteúdo está fundamentado na [Java Language Specification — Java SE 25](https://docs.oracle.com/javase/specs/jls/se25/html/index.html), a especificação oficial da linguagem publicada pela Oracle. Todas as demonstrações estão implementadas e podem ser executadas em ordem em um kernel Java para notebooks.

## Objetivos de aprendizagem

Ao final, você deverá ser capaz de:

- reconhecer os cinco tipos de tokens da gramática Java;
- escrever identificadores válidos e distinguir palavras reservadas de palavras contextuais;
- reconhecer as categorias de literais;
- identificar separadores e operadores no código-fonte;
- distinguir os oito tipos primitivos e os valores que representam;
- relacionar variáveis, declarações, expressões e instruções;
- decompor uma instrução Java em seus elementos constituintes.

## 1. Dos caracteres às instruções

O compilador transforma o texto de um programa em uma sequência de **tokens**. A gramática léxica da linguagem define cinco categorias de tokens:

1. identificadores;
2. palavras-chave;
3. literais;
4. separadores;
5. operadores.

A gramática sintática combina esses tokens para formar tipos, declarações, expressões e instruções. Espaços em branco e comentários são elementos de entrada, mas não são tokens.

**Referência oficial:** [JLS §3.5 — Input Elements and Tokens](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html#jls-3.5).

## 2. Identificadores

Um **identificador** é um nome usado para elementos do programa, como variáveis. Ele possui comprimento ilimitado, deve começar por uma letra Java e pode continuar com letras ou dígitos Java. A definição de “letra Java” inclui caracteres Unicode aceitos por `Character.isJavaIdentifierStart`.

Regras importantes:

- não pode começar com algarismo;
- não pode ser igual a uma palavra reservada, `true`, `false` ou `null`;
- diferencia maiúsculas de minúsculas: `total` e `Total` são nomes distintos;
- `_` sozinho é palavra reservada e não pode ser identificador;
- `$` é permitido por razões históricas, mas deve ser evitado em código escrito manualmente.

**Referência oficial:** [JLS §3.8 — Identifiers](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html#jls-3.8).

````java
int quantidadeDeAlunos = 30;
int QuantidadeDeAlunos = 25; // identificador diferente
double médiaDaTurma = 8.5;   // letras Unicode são permitidas
int valor2 = 2;              // dígito permitido após o início

System.out.printf("quantidade=%d, Quantidade=%d, média=%.1f, valor2=%d%n",
        quantidadeDeAlunos, QuantidadeDeAlunos, médiaDaTurma, valor2);

// Não compilam:
// int 2valor = 2;  // começa com algarismo
// int class = 1;   // class é palavra reservada
// int _ = 1;       // _ sozinho é palavra reservada
````

## 3. Palavras-chave

Palavras-chave possuem significado especial na gramática. A especificação diferencia:

- **palavras reservadas**, como `int`, `class`, `if`, `final`, `return` e `while`, que não podem ser usadas como identificadores;
- **palavras contextuais**, como `var`, `record`, `sealed` e `yield`, interpretadas como palavras-chave somente em contextos específicos.

`true` e `false` são literais booleanos, e `null` é o literal nulo; nenhum deles é palavra-chave. `const` e `goto` são reservadas, embora não sejam usadas pela linguagem.

**Referência oficial:** [JLS §3.9 — Keywords](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html#jls-3.9).

````java
int inteiroDemonstracao = 10; // int é palavra reservada
final int LIMITE_DEMONSTRACAO = 100; // final é palavra reservada
var valorInferido = 20;       // var é palavra contextual

System.out.printf("inteiro=%d, limite=%d, inferido=%d%n",
        inteiroDemonstracao, LIMITE_DEMONSTRACAO, valorInferido);
````

## 4. Literais

Um **literal** é a representação, no código-fonte, de um valor. A JLS define estas categorias:

| Categoria | Exemplos | Observação |
|---|---|---|
| inteiro | `10`, `0b1010`, `012`, `0xA`, `10L` | bases decimal, binária, octal e hexadecimal |
| ponto flutuante | `3.14`, `2.0F`, `1e3` | `double` por padrão; `F` indica `float` |
| booleano | `true`, `false` | são literais, não palavras-chave |
| caractere | `'A'`, `'\n'`, `'\u0041'` | representa uma unidade UTF-16 |
| string | `"Java"` | produz referência para um objeto `String` |
| bloco de texto | `"""..."""` | string que pode ocupar várias linhas |
| nulo | `null` | representa a referência nula |

O sublinhado pode separar dígitos para melhorar a leitura, como em `1_000_000`; ele não altera o valor.

**Referência oficial:** [JLS §3.10 — Literals](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html#jls-3.10).

````java
int decimalLiteral = 10;
int binarioLiteral = 0b1010;
int octalLiteral = 012;
int hexadecimalLiteral = 0xA;
long longLiteral = 10L;
float floatLiteral = 2.5F;
double doubleLiteral = 1.0e3;
boolean booleanLiteral = true;
char caractereLiteral = '\u0041';
String stringLiteral = "Java";
String textoLiteral = """
        primeira linha
        segunda linha
        """;
String referenciaNulaLiteral = null;

System.out.printf("inteiros: %d %d %d %d %d%n", decimalLiteral, binarioLiteral,
        octalLiteral, hexadecimalLiteral, longLiteral);
System.out.printf("reais: %.1f %.1f; booleano: %b; char: %c; string: %s%n",
        floatLiteral, doubleLiteral, booleanLiteral, caractereLiteral, stringLiteral);
System.out.print(textoLiteral);
System.out.println("referência nula? " + (referenciaNulaLiteral == null));
````

## 5. Separadores

Separadores são tokens que delimitam ou organizam partes do programa. A JLS lista treze separadores:

```text
(  )  {  }  [  ]  ;  ,  .  ...  @  ::
```

Seu significado depende da construção sintática. Por exemplo, `;` encerra uma declaração ou instrução, `,` separa declaradores ou argumentos, `{}` delimita um bloco, `[]` participa de tipos e acessos a arrays, e `.` qualifica um nome ou acessa um membro.

**Referência oficial:** [JLS §3.11 — Separators](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html#jls-3.11).

````java
int separadorA = 2, separadorB = 3; // vírgula e ponto e vírgula
int[] numerosSeparadores = {separadorA, separadorB}; // [], {} e vírgula
int somaSeparadores = (numerosSeparadores[0] + numerosSeparadores[1]); // () e []
System.out.println("Soma delimitada por separadores: " + somaSeparadores); // ponto e ()
````

## 6. Operadores

Operadores são tokens que participam da formação de expressões. Os principais grupos são:

| Grupo | Operadores comuns |
|---|---|
| aritméticos | `+`, `-`, `*`, `/`, `%` |
| incremento e decremento | `++`, `--` |
| relacionais | `<`, `<=`, `>`, `>=`, `instanceof` |
| igualdade | `==`, `!=` |
| lógicos condicionais | `&&`, `||` |
| bit a bit e deslocamento | `&`, `|`, `^`, `~`, `<<`, `>>`, `>>>` |
| atribuição | `=`, `+=`, `-=`, `*=`, `/=`, `%=` etc. |
| condicional | `? :` |
| criação e referências | `new`, `->`, `::` aparecem em construções específicas |

Precedência e associatividade determinam o agrupamento. Parênteses podem tornar esse agrupamento explícito. `&&` e `||` usam curto-circuito: o operando direito só é avaliado quando necessário.

**Referências oficiais:** [JLS §3.12 — Operators](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html#jls-3.12) e [JLS, capítulo 15 — Expressions](https://docs.oracle.com/javase/specs/jls/se25/html/jls-15.html).

````java
int operadorX = 7;
int operadorY = 2;
int somaOperadores = operadorX + operadorY;
int restoOperadores = operadorX % operadorY;
boolean comparacaoOperadores = operadorX > operadorY && operadorY != 0;
int deslocamentoOperadores = operadorY << 2;
String paridadeOperadores = operadorX % 2 == 0 ? "par" : "ímpar";
operadorX += 1;

System.out.printf("soma=%d, resto=%d, condição=%b, deslocamento=%d, paridade=%s, x=%d%n",
        somaOperadores, restoOperadores, comparacaoOperadores,
        deslocamentoOperadores, paridadeOperadores, operadorX);
````

## 7. Tipos primitivos

Java possui exatamente oito tipos primitivos. Eles não são objetos.

| Tipo | Categoria | Representação ou conjunto de valores |
|---|---|---|
| `byte` | inteiro com sinal | 8 bits: −128 a 127 |
| `short` | inteiro com sinal | 16 bits: −32.768 a 32.767 |
| `int` | inteiro com sinal | 32 bits: −2³¹ a 2³¹−1 |
| `long` | inteiro com sinal | 64 bits: −2⁶³ a 2⁶³−1 |
| `char` | inteiro sem sinal | unidade UTF-16 de 16 bits: `\u0000` a `\uFFFF` |
| `float` | ponto flutuante | IEEE 754 binário de 32 bits |
| `double` | ponto flutuante | IEEE 754 binário de 64 bits |
| `boolean` | lógico | `true` ou `false`; tamanho de armazenamento não é definido pela JLS |

Os inteiros usam complemento de dois. Tipos de ponto flutuante também incluem zeros positivo e negativo, infinitos e `NaN`. `boolean` não é um tipo numérico. `String` é tipo por referência, mesmo quando seu valor é escrito com um literal.

**Referências oficiais:** [JLS §4.2 — Primitive Types and Values](https://docs.oracle.com/javase/specs/jls/se25/html/jls-4.html#jls-4.2), [§4.2.1 — Integral Types and Values](https://docs.oracle.com/javase/specs/jls/se25/html/jls-4.html#jls-4.2.1), [§4.2.3 — Floating-Point Types](https://docs.oracle.com/javase/specs/jls/se25/html/jls-4.html#jls-4.2.3) e [§4.2.5 — The boolean Type](https://docs.oracle.com/javase/specs/jls/se25/html/jls-4.html#jls-4.2.5).

````java
byte byteDemo = 127;
short shortDemo = 32_767;
int intDemo = 2_147_483_647;
long longDemo = 9_223_372_036_854_775_807L;
char charDemo = '\uFFFF';
float floatDemo = 3.5F;
double doubleDemo = 3.5;
boolean booleanDemo = true;

System.out.printf("byte=%d, short=%d, int=%d, long=%d%n",
        byteDemo, shortDemo, intDemo, longDemo);
System.out.printf("char=%d, float=%.1f, double=%.1f, boolean=%b%n",
        (int) charDemo, floatDemo, doubleDemo, booleanDemo);
````

## 8. Variáveis

Uma **variável** é um local de armazenamento associado a um tipo. O tipo determina quais valores a variável pode conter e quais operações são permitidas. A JLS classifica variáveis em variáveis de classe, variáveis de instância, componentes de array, parâmetros e variáveis locais.

Neste notebook usamos variáveis locais do ambiente interativo. Cada uma possui:

- um tipo, explícito ou inferido com `var`;
- um nome identificado por um identificador;
- um valor compatível com o tipo;
- um escopo no qual pode ser referenciada.

Uma variável `final` pode receber valor apenas uma vez. Isso cria uma variável não reatribuível; não cria um novo tipo.

**Referência oficial:** [JLS §4.12 — Variables](https://docs.oracle.com/javase/specs/jls/se25/html/jls-4.html#jls-4.12).

````java
int saldoVariavel = 100;        // tipo explícito, nome e valor inicial
saldoVariavel = 125;            // atribuição de novo valor
var quantidadeVariavel = 3;     // tipo int inferido do inicializador
final double TAXA_VARIAVEL = 0.10;

{
    int variavelDoBloco = 5;    // escopo limitado ao bloco
    saldoVariavel += variavelDoBloco;
}

System.out.printf("saldo=%d, quantidade=%d, taxa=%.2f%n",
        saldoVariavel, quantidadeVariavel, TAXA_VARIAVEL);
// TAXA_VARIAVEL = 0.20;        // erro: variável final já inicializada
// System.out.println(variavelDoBloco); // erro: fora do escopo
````

## 9. Declarações

Uma **declaração** introduz uma entidade no programa e pode associar a ela um nome. Para variáveis locais, a forma geral é:

```java
[modificadores] tipo identificador [= inicializador] [, identificador [= inicializador] ...];
```

A declaração `int quantidade = 2;` combina o tipo `int`, o identificador `quantidade`, o operador de atribuição `=`, o literal `2` e o separador `;`. Declarar não é o mesmo que atribuir: a declaração introduz a variável; uma atribuição posterior apenas muda seu valor.

`var` pode substituir o tipo em certas declarações de variáveis locais quando há um inicializador do qual o compilador infere o tipo. Ele não torna Java uma linguagem dinamicamente tipada.

**Referências oficiais:** [JLS §6.1 — Declarations](https://docs.oracle.com/javase/specs/jls/se25/html/jls-6.html#jls-6.1) e [JLS §14.4 — Local Variable Declaration Statements](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.4).

````java
int declaracaoSemInicializador;
declaracaoSemInicializador = 40;
int primeiraDeclaracao = 1, segundaDeclaracao = 2;
final int DECLARACAO_FINAL = 3;
var declaracaoInferida = 4;

System.out.println(declaracaoSemInicializador + primeiraDeclaracao
        + segundaDeclaracao + DECLARACAO_FINAL + declaracaoInferida);
````

## 10. Expressões

Uma **expressão** é uma construção formada por variáveis, literais, operadores e outras expressões que, ao ser avaliada, produz um valor, uma variável ou nenhuma informação útil (`void`).

Exemplos:

- `10` — expressão literal;
- `quantidade` — expressão de nome;
- `quantidade * 20` — expressão multiplicativa;
- `saldo = 100` — expressão de atribuição;
- `idade >= 18` — expressão relacional que produz `boolean`;
- `condicao ? valor1 : valor2` — expressão condicional.

A precedência define como uma expressão sem parênteses é agrupada. Java avalia os operandos da esquerda para a direita, mas alguns operadores, como `&&`, `||` e `?:`, avaliam apenas os operandos necessários.

**Referência oficial:** [JLS, capítulo 15 — Expressions](https://docs.oracle.com/javase/specs/jls/se25/html/jls-15.html).

````java
int quantidadeExpressao = 4;
int precoExpressao = 25;
int totalExpressao = quantidadeExpressao * precoExpressao;
boolean descontoExpressao = totalExpressao >= 100;
int totalComDescontoExpressao = descontoExpressao ? totalExpressao - 10 : totalExpressao;

System.out.printf("total=%d, desconto=%b, total final=%d%n",
        totalExpressao, descontoExpressao, totalComDescontoExpressao);
````

## 11. Instruções

Uma **instrução** (*statement*) representa uma ação ou controla a execução. O capítulo 14 da JLS define, entre outras, estas formas:

- bloco: `{ ... }`;
- instrução vazia: `;`;
- declaração de variável local: `int x = 1;`;
- instrução de expressão: `x++;` ou `x = 2;`;
- instruções de seleção, repetição e desvio.

Nem toda expressão pode aparecer sozinha como instrução. Atribuições, incrementos/decrementos, criação de objetos e invocações são exemplos de expressões permitidas como instruções. Um bloco agrupa zero ou mais instruções e cria um novo escopo.

**Referências oficiais:** [JLS, capítulo 14 — Blocks, Statements, and Patterns](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html), [§14.2 — Blocks](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.2) e [§14.8 — Expression Statements](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html#jls-14.8).

````java
int contadorInstrucao = 0;   // instrução de declaração local
contadorInstrucao = 1;       // instrução de expressão: atribuição
contadorInstrucao++;         // instrução de expressão: incremento
;                            // instrução vazia
{                            // início de uma instrução de bloco
    int incrementoInstrucao = 3;
    contadorInstrucao += incrementoInstrucao;
}
if (contadorInstrucao == 5) { // instrução de seleção
    System.out.println("contador chegou a 5");
}
````

## 12. Demonstração integrada

Considere a instrução:

```java
int total = quantidade * preco;
```

Ela reúne os elementos estudados:

| Trecho | Papel |
|---|---|
| `int` | palavra-chave e nome de tipo primitivo |
| `total`, `quantidade`, `preco` | identificadores |
| `=` e `*` | operadores |
| `;` | separador |
| `int total = ...` | declaração de variável |
| `quantidade * preco` | expressão multiplicativa |
| a linha completa | instrução de declaração de variável local |

Um literal pode participar da mesma construção, como em `int total = quantidade * 25;`.

````java
int quantidadeIntegracao = 3;
int precoIntegracao = 25;
int totalIntegracao = quantidadeIntegracao * precoIntegracao;
System.out.printf("%d unidades × R$ %d = R$ %d%n",
        quantidadeIntegracao, precoIntegracao, totalIntegracao);
````

## 13. Erros comuns

1. **Usar palavra reservada como identificador:** `int class = 1;` não compila.
2. **Confundir `true`, `false` e `null` com palavras-chave:** eles são literais.
3. **Confundir literal com variável:** `10` representa um valor; `quantidade` nomeia uma variável.
4. **Acreditar que `String` é primitivo:** é um tipo por referência.
5. **Omitir o sufixo de `float`:** `2.5` é `double`; para `float`, escreva `2.5F`.
6. **Esperar parte decimal de divisão inteira:** `5 / 2` produz `2`.
7. **Acreditar que `var` muda de tipo:** o tipo é inferido na declaração e permanece estático.
8. **Ler variável local antes de inicializá-la:** a análise de atribuição definida do compilador rejeita o uso.
9. **Confundir expressão com instrução:** uma expressão produz um resultado; uma instrução representa uma ação ou controle.
10. **Ignorar escopo:** uma variável declarada em um bloco não existe fora dele.

## Síntese

- **Literais** representam valores diretamente no código-fonte.
- **Identificadores** nomeiam elementos do programa.
- **Palavras-chave** possuem significado definido pela gramática.
- **Separadores** delimitam e organizam construções.
- **Operadores** participam da formação de expressões.
- **Tipos primitivos** definem os valores fundamentais não referenciais.
- **Variáveis** são locais de armazenamento associados a tipos.
- **Declarações** introduzem entidades e nomes.
- **Expressões** são avaliadas e produzem resultados.
- **Instruções** representam ações e controle de execução.
