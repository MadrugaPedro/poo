## 1. Da unidade de compilação à execução

Em um arquivo-fonte tradicional, uma **unidade de compilação** pode conter declaração de pacote, importações e declarações de tipos. Uma classe pode reunir campos, construtores e métodos. A JVM inicia uma aplicação convencional invocando o método `main`.

```java
package exemplo;                 // pacote, se houver

import java.time.LocalDate;       // importação

public class Ola {                // declaração de tipo
    public static void main(String[] args) {
        System.out.println("Olá em " + LocalDate.now());
    }
}
```

A JLS define a linguagem; a JVM e suas instruções são especificadas separadamente. Compilar (`javac`) normalmente transforma fonte `.java` em bytecode `.class`; executar (`java`) inicia a JVM. Em um notebook/JShell, essa infraestrutura fica escondida para favorecer experimentação.

**Referências:** [JLS §7.3 — Compilation Units](https://docs.oracle.com/javase/specs/jls/se25/html/jls-7.html#jls-7.3) e [JLS §12.1.4 — Invoke `main`](https://docs.oracle.com/javase/specs/jls/se25/html/jls-12.html#jls-12.1.4).

````java
import java.util.Scanner;

System.out.println("Olá, Java!");
System.out.printf("Versão da JVM: %s%n", System.getProperty("java.version"));
````

## 2. Estrutura léxica

Antes de analisar classes e expressões, o compilador transforma caracteres Unicode em **tokens**. Entre os tokens estão identificadores, palavras-chave, literais, separadores e operadores. Espaços em branco e comentários normalmente apenas separam tokens.

- **Identificador:** nome atribuído pelo programador, como `saldoAtual`. Java diferencia maiúsculas de minúsculas.
- **Palavra-chave:** termo reservado ou contextual, como `class`, `int`, `return` e `var`.
- **Literal:** representação direta de um valor, como `42`, `3.14`, `'J'`, `true`, `"Java"` e `null`.
- **Separador:** símbolo estrutural, como `(`, `)`, `{`, `}`, `[`, `]`, `;`, `,` e `.`.
- **Comentário:** `// linha`, `/* bloco */` ou `/** documentação */`.

Convenções não são regras gramaticais: classes costumam usar `PascalCase`, variáveis e métodos `camelCase`, e constantes `MAIUSCULAS_COM_UNDERLINE`.

**Referência:** [JLS, capítulo 3 — Lexical Structure](https://docs.oracle.com/javase/specs/jls/se25/html/jls-3.html).

````java
int decimal = 1_000_000;        // _ melhora a leitura
int binario = 0b1010;           // 10
int hexadecimal = 0xFF;         // 255
long populacao = 8_000_000_000L;
double cientifico = 6.022e23;
float taxa = 2.5F;
char inicial = 'J';
String linguagem = "Java";
String blocoDeTexto = """
        primeira linha
        segunda linha
        """;
System.out.printf("%d, %d, %d, %.3e, %c, %s%n",
        decimal, binario, hexadecimal, cientifico, inicial, linguagem);
System.out.println(blocoDeTexto);
````

## 3. Tipos, valores e variáveis

Java é estaticamente tipada: toda variável e toda expressão possui um tipo determinado em compilação. Há dois grandes grupos.

### Tipos primitivos

| Tipo | Natureza | Tamanho | Exemplo |
|---|---|---:|---|
| `byte` | inteiro com sinal | 8 bits | `byte b = 100;` |
| `short` | inteiro com sinal | 16 bits | `short s = 30_000;` |
| `int` | inteiro com sinal | 32 bits | `int n = 42;` |
| `long` | inteiro com sinal | 64 bits | `long n = 42L;` |
| `char` | unidade de código UTF-16 sem sinal | 16 bits | `char c = 'A';` |
| `float` | ponto flutuante IEEE 754 | 32 bits | `float f = 1.5F;` |
| `double` | ponto flutuante IEEE 754 | 64 bits | `double d = 1.5;` |
| `boolean` | lógico | não fixado pela JLS | `boolean ok = true;` |

Tipos inteiros têm faixas exatas e aritmética sujeita a *overflow*. Ponto flutuante inclui infinito, zero positivo/negativo e `NaN`; não representa exatamente todo decimal. `boolean` não é um número e não se converte para `int`.

### Tipos de referência

Classes, interfaces e arrays definem tipos de referência. A variável guarda uma referência a um objeto (ou `null`), não o objeto em si. `String` é classe e seus objetos são imutáveis. Arrays também são objetos.

**Referência:** [JLS, capítulo 4 — Types, Values, and Variables](https://docs.oracle.com/javase/specs/jls/se25/html/jls-4.html).

````java
System.out.printf("byte:  %d a %d%n", Byte.MIN_VALUE, Byte.MAX_VALUE);
System.out.printf("int:   %d a %d%n", Integer.MIN_VALUE, Integer.MAX_VALUE);
System.out.printf("long:  %d a %d%n", Long.MIN_VALUE, Long.MAX_VALUE);
System.out.printf("char:  %d a %d%n", (int) Character.MIN_VALUE, (int) Character.MAX_VALUE);
System.out.println("1.0 / 0.0 = " + (1.0 / 0.0));
System.out.println("0.0 / 0.0 = " + (0.0 / 0.0));
System.out.println("0.1 + 0.2 = " + (0.1 + 0.2));
````

## 4. Declaração, inicialização, `var` e `final`

Uma declaração introduz nome e tipo; o inicializador fornece o primeiro valor. Variáveis locais devem estar **definitivamente atribuídas** antes da leitura. Campos, componentes de array e parâmetros possuem regras diferentes; campos e componentes recebem valores padrão, mas variáveis locais não.

`var` pede ao compilador que infira o tipo de uma variável local a partir do inicializador. Não cria tipagem dinâmica e não pode ser usado sem inicializador, em campos ou como tipo de retorno.

Uma variável `final` só pode ser atribuída uma vez. Se ela contém uma referência, a referência não muda, mas o objeto referenciado ainda pode ser mutável.

````java
int quantidade;
quantidade = 3;
var produto = "Caderno";     // inferido como String
final double PRECO_UNITARIO = 12.50;
double total = quantidade * PRECO_UNITARIO;
System.out.printf("%d x %s = R$ %.2f%n", quantidade, produto, total);
````

### Valor primitivo × referência

Atribuir um primitivo copia seu valor. Atribuir uma referência copia a referência; duas variáveis podem então apontar para o mesmo objeto.

````java
int original = 10;
int copia = original;
copia++;
System.out.printf("primitivos: original=%d, copia=%d%n", original, copia);

int[] primeiraReferencia = {10, 20};
int[] segundaReferencia = primeiraReferencia;
segundaReferencia[0] = 99;
System.out.printf("mesmo array: %d e %d%n",
        primeiraReferencia[0], segundaReferencia[0]);
````

## 5. Conversões e contextos

O contexto determina quais conversões são permitidas. As mais importantes neste ponto são:

- **ampliação (*widening*):** costuma ocorrer implicitamente, como `int` para `long`; pode perder precisão ao chegar a ponto flutuante;
- **estreitamento (*narrowing*):** exige *cast*, como `double` para `int`, e pode perder magnitude ou precisão;
- **promoção numérica:** operadores promovem operandos menores que `int` para `int`;
- **boxing/unboxing:** conversão entre primitivos e classes empacotadoras, como `int` e `Integer`;
- **conversão para `String`:** aparece, por exemplo, na concatenação com `+`.

O *cast* declara intenção, mas não torna a conversão segura.

**Referência:** [JLS, capítulo 5 — Conversions and Contexts](https://docs.oracle.com/javase/specs/jls/se25/html/jls-5.html).

````java
int inteiro = 1_000_000_001;
long ampliado = inteiro;
float ampliadoComPerda = inteiro;
double real = 9.99;
int truncado = (int) real;
byte estourado = (byte) 130;
Integer empacotado = inteiro;       // boxing
int desempacotado = empacotado;     // unboxing
System.out.printf("long=%d, float=%.0f, truncado=%d, byte=%d, unboxed=%d%n",
        ampliado, ampliadoComPerda, truncado, estourado, desempacotado);
````

## 6. Operadores e expressões

Uma expressão é avaliada e produz um valor, uma variável ou nenhuma informação útil (`void`). Principais grupos:

| Grupo | Operadores comuns |
|---|---|
| aritméticos | `+`, `-`, `*`, `/`, `%` |
| unários | `+`, `-`, `++`, `--`, `!`, `~` |
| relacionais | `<`, `<=`, `>`, `>=`, `instanceof` |
| igualdade | `==`, `!=` |
| lógicos condicionais | `&&`, `||` |
| bit a bit | `&`, `|`, `^`, `~`, `<<`, `>>`, `>>>` |
| atribuição | `=`, `+=`, `-=`, `*=`, `/=`, `%=` etc. |
| condicional | `condição ? valor1 : valor2` |

Precedência define o agrupamento, não a ordem temporal de avaliação. Java avalia operandos da esquerda para a direita. `&&` e `||` usam curto-circuito: o operando direito só é avaliado se necessário. Prefira parênteses quando eles tornam a intenção mais clara.

Na divisão inteira, a parte fracionária é descartada. Divisão inteira por zero lança `ArithmeticException`; divisão de ponto flutuante por zero segue IEEE 754.

**Referência:** [JLS, capítulo 15 — Expressions](https://docs.oracle.com/javase/specs/jls/se25/html/jls-15.html).

````java
int a = 7;
int b = 2;
System.out.println("divisão inteira: " + a / b);
System.out.println("divisão real:    " + (double) a / b);
System.out.println("resto:            " + a % b);
System.out.println("precedência:      " + (2 + 3 * 4));
System.out.println("com parênteses:   " + ((2 + 3) * 4));
boolean denominadorValido = b != 0;
boolean divisaoExata = denominadorValido && a % b == 0;
String paridade = a % 2 == 0 ? "par" : "ímpar";
System.out.printf("exata=%b; %d é %s%n", divisaoExata, a, paridade);
````

### Igualdade: `==` não substitui `equals`

Para primitivos, `==` compara valores após as conversões aplicáveis. Para referências, compara se elas apontam para o mesmo objeto. Classes como `String` implementam `equals` para comparar conteúdo. O *pool* de strings pode fazer certos literais compartilharem uma referência, portanto usar `==` para conteúdo gera código frágil.

````java
String texto1 = new String("Java");
String texto2 = new String("Java");
System.out.println("mesma referência? " + (texto1 == texto2));
System.out.println("mesmo conteúdo?   " + texto1.equals(texto2));

String possivelmenteNulo = null;
System.out.println("comparação segura: " + "Java".equals(possivelmenteNulo));
````

### Incremento e overflow

`x++` produz o valor antigo e depois incrementa; `++x` incrementa e produz o novo valor. Em instruções isoladas ambos atualizam `x`, mas dentro de expressões podem confundir. Inteiros não sinalizam overflow automaticamente.

````java
int limite = Integer.MAX_VALUE;
int comOverflow = limite + 1;
System.out.printf("%d + 1 = %d%n", limite, comOverflow);

int contador = 5;
System.out.printf("pós=%d; agora=%d%n", contador++, contador);
System.out.printf("pré=%d; agora=%d%n", ++contador, contador);
````

## 7. Instruções, blocos e escopo

Uma **expressão** calcula algo; uma **instrução** representa uma ação completa; um **bloco** agrupa instruções entre `{` e `}`. O escopo delimita onde um nome pode ser usado. Uma variável local declarada dentro de um bloco não existe fora dele.

Nem toda expressão pode aparecer sozinha como instrução. Chamadas de método, atribuições, incrementos/decrementos e criação de objetos são exemplos permitidos. O ponto e vírgula termina muitas instruções, não blocos.

**Referências:** [JLS §6.3 — Scope of a Declaration](https://docs.oracle.com/javase/specs/jls/se25/html/jls-6.html#jls-6.3) e [JLS, capítulo 14 — Blocks, Statements, and Patterns](https://docs.oracle.com/javase/specs/jls/se25/html/jls-14.html).

````java
int nivelExterno = 10;
{
    int nivelInterno = 20;
    System.out.println(nivelExterno + nivelInterno);
}
// System.out.println(nivelInterno); // erro: fora do escopo
System.out.println(nivelExterno);
````

## 8. Métodos

Métodos dão nome a comportamentos e favorecem reúso. Uma declaração informa modificadores, tipo de retorno, nome, parâmetros e corpo. `void` indica ausência de valor de retorno. Em Java, os argumentos são sempre passados **por valor**: para uma referência, o valor copiado é a própria referência.

A assinatura considerada para sobrecarga é formada pelo nome e pelos tipos dos parâmetros; somente mudar o retorno não cria outra sobrecarga.

**Referência:** [JLS §8.4 — Method Declarations](https://docs.oracle.com/javase/specs/jls/se25/html/jls-8.html#jls-8.4).

````java
static double calcularMedia(double nota1, double nota2) {
    return (nota1 + nota2) / 2.0;
}

static String classificar(double media) {
    return media >= 6.0 ? "aprovado" : "em recuperação";
}

double media = calcularMedia(7.5, 8.0);
System.out.printf("Média %.1f: %s%n", media, classificar(media));
````

## 9. Entrada e saída

Entrada/saída pertence à biblioteca padrão, não à sintaxe nuclear da linguagem. `System.out` oferece saída; `Scanner` é uma opção didática para ler entrada formatada. Em programas reais, valide dados e feche recursos que você abriu. Evite fechar um `Scanner` ligado a `System.in` no meio de uma aplicação, pois isso também fecha a entrada padrão.

A célula abaixo simula a entrada para continuar reprodutível. Troque o `Scanner` por `new Scanner(System.in)` em uma aplicação interativa.

````java
String entradaSimulada = "Ada 9.5 8.0";
try (Scanner leitor = new Scanner(entradaSimulada)) {
    String nome = leitor.next();
    double n1 = leitor.nextDouble();
    double n2 = leitor.nextDouble();
    double resultado = (n1 + n2) / 2.0;
    System.out.printf("%s: média %.2f%n", nome, resultado);
}
````

## 10. Estudo de caso — cálculo de compra

O exemplo integra constantes, tipos, conversões, operadores, métodos, validação e formatação. Observe a escolha de `long` para centavos: dinheiro não deve ser calculado com `double` quando exatidão decimal é requisito. Em sistemas maiores, `BigDecimal` é outra escolha comum.

````java
static long calcularSubtotal(long precoEmCentavos, int quantidade) {
    if (precoEmCentavos < 0 || quantidade < 0) {
        throw new IllegalArgumentException("Preço e quantidade devem ser não negativos");
    }
    return Math.multiplyExact(precoEmCentavos, quantidade);
}

static String formatarReais(long centavos) {
    return "R$ %d,%02d".formatted(centavos / 100, Math.abs(centavos % 100));
}

final long PRECO_CADERNO = 1_250;
int unidades = 3;
long subtotal = calcularSubtotal(PRECO_CADERNO, unidades);
boolean freteGratis = subtotal >= 3_000;
long frete = freteGratis ? 0 : 500;
long totalCompra = Math.addExact(subtotal, frete);
System.out.printf("Subtotal: %s%nFrete: %s%nTotal: %s%n",
        formatarReais(subtotal), formatarReais(frete), formatarReais(totalCompra));
````

## 11. Erros comuns e como raciocinar sobre eles

1. **Esperar decimal de `5 / 2`:** ambos os operandos são `int`; converta um deles para `double`.
2. **Usar `==` para conteúdo de `String`:** use `equals` ou `Objects.equals` quando houver valores nulos.
3. **Acreditar que `var` muda de tipo:** o tipo é inferido uma vez e continua estático.
4. **Ignorar overflow inteiro:** use faixa adequada e, quando necessário, `Math.addExact`/`multiplyExact`.
5. **Usar `double` para dinheiro:** a representação binária pode introduzir arredondamentos; use centavos inteiros ou `BigDecimal`.
6. **Fazer *cast* sem analisar perda:** estreitamento pode truncar, arredondar ou descartar bits.
7. **Confundir `final` com objeto imutável:** `final` impede trocar a referência, não alterar o objeto.
8. **Ler variável local antes da atribuição:** o compilador rejeita o uso.
9. **Misturar `nextInt()` e `nextLine()` sem consumir a quebra de linha:** entenda o delimitador do `Scanner`.
